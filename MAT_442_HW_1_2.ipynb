## 1.2.1 Linear Spaces

A set of vectors is **linearly independent** if no vector in the set can be written as a combination of the others.
We can check this using the **rank** of the matrix formed by the vectors: if the rank equals the number of vectors, they are independent and form a **basis** for their span.

```python
# Three vectors in R^3
v1 = np.array([1, 5, 7])
v2 = np.array([2, 1, 0])
v3 = np.array([0, 11, 101])   # note: v3 = 2*v2 - v1... let's check dependency

M = np.array([v1, v2, v3])
rank = np.linalg.matrix_rank(M)

print("Matrix of vectors:\n", M)
print("Rank:", rank)

if rank == M.shape[0]:
    print("The vectors are linearly independent and form a basis for their span.")
else:
    print("The vectors are linearly dependent — they do not form a basis.")
```

## 1.2.2 Orthogonality

Two vectors **u** and **v** are orthogonal if their dot product is zero: u · v = 0.
The **projection** of a vector **b** onto a vector **a** is given by:

proj_a(b) = ( (a · b) / (a · a) ) * a

We'll check orthogonality between two vectors, then compute a projection.

```python
a = np.array([3, 0, -1])
b = np.array([1, 1, 1])

dot_ab = np.dot(a, b)
print(f"a · b = {dot_ab}")
print("Orthogonal!" if dot_ab == 0 else "Not orthogonal.")

# Projection of a new vector c onto a
c = np.array([2, 3, 1])
proj_c_on_a = (np.dot(a, c) / np.dot(a, a)) * a
print("\nProjection of c onto a:", proj_c_on_a)

# Verify: (c - proj) should be orthogonal to a
residual = c - proj_c_on_a
print("Residual · a =", np.dot(residual, a), "(should be ~0)")
```

## 1.2.3 Gram–Schmidt Process

Gram–Schmidt converts a set of linearly independent vectors into an **orthonormal** set (mutually orthogonal, unit length), by subtracting off projections onto previous vectors, then normalizing.

We implement it manually here, then verify orthonormality, and cross-check with NumPy's QR decomposition.

```python
def gram_schmidt(vectors):
    basis = []
    for v in vectors:
        w = v.astype(float).copy()
        for u in basis:
            w -= (np.dot(v, u) / np.dot(u, u)) * u
        u_normalized = w / np.linalg.norm(w)
        basis.append(u_normalized)
    return np.array(basis)

vectors = [np.array([1, 1, 0]), np.array([1, 0, 1]), np.array([0, 1, 1])]
ortho_basis = gram_schmidt(vectors)

print("Orthonormal basis:\n", ortho_basis)

# Verify orthogonality (dot products between different vectors ~0)
print("\nDot products (should be ~0 off-diagonal, ~1 on diagonal):")
print(ortho_basis @ ortho_basis.T)

# Cross-check with numpy's QR decomposition
Q, R = np.linalg.qr(np.array(vectors).T.astype(float))
print("\nNumPy QR orthonormal columns (Q):\n", Q.T)
```

## 1.2.4 Eigenvalues and Eigenvectors

For a square matrix A, an eigenvector **v** and eigenvalue λ satisfy:

A v = λ v

We compute both using NumPy, then verify the equation directly.

```python
A = np.array([[4, 1],
              [2, 3]])

eigenvalues, eigenvectors = np.linalg.eig(A)

print("Eigenvalues:", eigenvalues)
print("Eigenvectors (columns):\n", eigenvectors)

# Verify A v = lambda v for the first eigenpair
v = eigenvectors[:, 0]
lam = eigenvalues[0]

Av = A @ v
lambda_v = lam * v

print("\nA @ v      =", Av)
print("lambda * v =", lambda_v)
print("Match:", np.allclose(Av, lambda_v))
```
